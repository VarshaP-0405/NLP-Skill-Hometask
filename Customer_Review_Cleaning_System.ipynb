```python
#Load the IMDb 50K and review dataset using Pandas
import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')

df=pd.read_csv("IMDB_Dataset_CLEANED.csv",engine="python")
print(df.head())
print(df.columns)
print(df.shape)

# Inspect missing values and duplicate reviews
print("Total missing values:",df.isnull().sum())
print("Number of duplicate reviews:",df['review'].duplicated().sum())
df=df.dropna(subset=['review'])
df=df.drop_duplicates(subset=['review'])
print(df.shape)

#Convert text to lowercase
df['lowercase_review']=df['review'].str.lower()
print(df[['review','lowercase_review']].head())
#Remove HTML tags, URLs,punctuation,numbers, and special characters using Regular Expressions
df['clean_review']=df['lowercase_review'].apply(lambda x: re.sub(r'<.*?','',x))
print(df[['review','clean_review']].head())

df['clean_review']=df['clean_review'].apply(lambda x: re.sub(r'http\S+','',x))
df['clean_review'] = df['clean_review'].apply(
    lambda x: re.sub(r'[^a-zA-Z\s]', '', x)
)
df['clean_review']=df['clean_review'].apply(lambda x: re.sub(r'\s+',' ',x).strip())

#Remove stopwords using NLTK
stop_words = set(stopwords.words('english'))
print(list(stop_words)[:20])
df['clean_review'] = df['clean_review'].apply(
    lambda x: ' '.join(
        word for word in x.split()
        if word not in stop_words
    )
)

#Tokenise the reviews
df['tokens']=df['clean_review'].apply(word_tokenize)
print(df[['clean_review','tokens']].head())

#Apply Stemming or lemmatisation
lemmatizer=WordNetLemmatizer()
df['lemmatized_tokens']=df['tokens'].apply( lambda tokens: [lemmatizer.lemmatize(word) for word in tokens])
print(df[['tokens', 'lemmatized_tokens']].head())

#Store the cleaned reviews in a new column
df['cleaned_review'] = df['lemmatized_tokens'].apply(
    lambda x: ' '.join(x)
)
#compare original and cleaned reviews
for i in range(5):
    print("ORIGINAL:")
    print(df['review'].iloc[i])

    print("\nCLEANED:")
    print(df['cleaned_review'].iloc[i])

    print("-" * 80)
#export the preprocessed dataset
df.to_csv("IMDB_preprocessed.csv",index=False)
print("Preprocessed dataset saved successfully!")
```
